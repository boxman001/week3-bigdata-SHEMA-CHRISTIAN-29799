```python
total = 0
valid = 0
bad = 0
bad_ids = []

with open("week3_students.csv", "r") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")

    student_id = parts[0]

    try:
        score = int(parts[7])
        total += score
        valid += 1

    except ValueError:
        bad += 1
        bad_ids.append(student_id)

average = total / valid

print("Valid:", valid)
print("Bad:", bad)
print("Bad IDs:", bad_ids)
print("Average:", round(average, 1))
```

```python
passed = 0
failed = 0

best_score = -1
best_name = ""

with open("week3_students.csv", "r") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")

    name = parts[1]

    try:
        score = int(parts[7])

        if score >= 50:
            passed += 1
        else:
            failed += 1

        if score > best_score:
            best_score = score
            best_name = name

    except ValueError:
        continue

print("Passed:", passed)
print("Failed:", failed)
print("Top student:", best_name)
print("Top score:", best_score)
print("Top student's grade:", get_grade(best_score))
```

```python
sums = {}
counts = {}

with open("week3_students.csv", "r") as f:
    lines = f.readlines()

for line in lines[1:]:
    parts = line.strip().split(",")

    district = parts[5]

    try:
        score = int(parts[7])

        sums[district] = sums.get(district, 0) + score
        counts[district] = counts.get(district, 0) + 1

    except ValueError:
        continue


highest_district = ""
highest_average = -1

for district in sums:
    average = sums[district] / counts[district]

    print(district, "average:", round(average, 1))

    if average > highest_average:
        highest_average = average
        highest_district = district

print("Highest average district:", highest_district)
print("Highest average:", round(highest_average, 1))
```
